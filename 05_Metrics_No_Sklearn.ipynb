# Metrics from scratch

Accuracy, per-class precision/recall/F1 and confusion matrix without scikit-learn. Replace the demo predictions with your model's predictions.

**No scikit-learn is used.** Requirements: `pandas`, `openpyxl`, `numpy`, `matplotlib`. Put `data.xlsx` beside the notebook. Change `TARGET` if the last column is not your class label.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from math import log2

FILE = "data.xlsx"
df = pd.read_excel(FILE).dropna().copy()
print(df.head())
print("Columns:", list(df.columns))

# Basic evaluation metrics without scikit-learn.
# For a real evaluation, create a train/test split before fitting your model.
TARGET = df.columns[-1]
y_true = df[TARGET].tolist()

# Replace this demo prediction list with predictions from your algorithm.
# This placeholder uses the majority class so the notebook runs as-is.
majority = Counter(y_true).most_common(1)[0][0]
y_pred = [majority] * len(y_true)

labels = sorted(set(y_true) | set(y_pred), key=str)
accuracy = sum(a == b for a, b in zip(y_true, y_pred)) / len(y_true)

print("Accuracy:", round(accuracy, 3))
print("\\nClass-wise metrics:")
for lab in labels:
    tp = sum(a == lab and b == lab for a, b in zip(y_true, y_pred))
    fp = sum(a != lab and b == lab for a, b in zip(y_true, y_pred))
    fn = sum(a == lab and b != lab for a, b in zip(y_true, y_pred))
    precision = tp/(tp+fp) if tp+fp else 0
    recall = tp/(tp+fn) if tp+fn else 0
    f1 = 2*precision*recall/(precision+recall) if precision+recall else 0
    print(f"{lab}: precision={precision:.3f}, recall={recall:.3f}, F1={f1:.3f}")

cm = pd.DataFrame(0, index=labels, columns=labels)
for a, b in zip(y_true, y_pred):
    cm.loc[a, b] += 1
print("\\nConfusion matrix (rows=actual, columns=predicted):")
display(cm)
plt.figure(figsize=(6,5))
plt.imshow(cm.values)
plt.xticks(range(len(labels)), labels, rotation=45)
plt.yticks(range(len(labels)), labels)
plt.xlabel("Predicted"); plt.ylabel("Actual")
plt.title("Confusion matrix")
for i in range(len(labels)):
    for j in range(len(labels)):
        plt.text(j, i, cm.iloc[i, j], ha="center", va="center")
plt.tight_layout(); plt.show()
```
